```python

# 1. Convert the column to numbers.
# errors='coerce' turns any text (like '1.2k') into NaN so it doesn't crash.
df["rating_count"] = pd.to_numeric(df["rating_count"], errors="coerce")
# 2. Now fill the NaN values with the mean
df["rating_count"].fillna(df["rating_count"].mean(), inplace=True)
```

```python
print(df["rating_count"])
```

```python
df.isnull().sum()
```

```python
# Convert the column to string type first
df["discounted_price"] = df["discounted_price"].astype(str)
df['actual_price'] = df['actual_price'].astype(str)

# Then apply the replace and conversion
df["discounted_price"] = df["discounted_price"].str.replace("₹", "").str.replace(",", "").astype(float)
df['actual_price'] = df['actual_price'].str.replace("₹", "").str.replace(",", "").astype(float)
```

```python
# Convert to string first
df['discount_percentage'] = df['discount_percentage'].astype(str)

# Then remove '%' and convert to float
df['discount_percentage'] = df['discount_percentage'].str.replace('%', '').astype(float)
df['discount_percentage']
```

```python
#clean Rating :some rows might have invalid chars like '|'replace them
# Convert to string first
df['rating']= df['rating'].astype(str)
df['rating_count']= df['rating_count'].astype(str)
df['rating']=pd.to_numeric(df['rating'],errors='coerce')
df['rating_count']=df['rating_count'].str.replace(',','').astype(float)
df['rating_count']
```

```python
df.head()
```

```python
df.drop(columns=['discount_percentage','user_id',	'user_name',	'review_id',	'review_title','review_content','img_link','product_link'],axis=1,inplace=True)
```

```python
df.drop(columns=[	'product_id','product_name'],axis=1,inplace=True)
```

```python
df.drop(columns=[	'about_product'],axis=1,inplace=True)
```

```python
df.head()
```

```python
df.shape
```

```python
df['category']=df['category'].str.split('|').str[0]
```

```python
df=pd.get_dummies(df,columns=['category'],drop_first=True,dtype=int)
```

```python
df.head()
```

```python
x=df.drop('rating',axis=1)
y=df['rating']
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train.shape)
print(y_train.shape)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
#Define the model of ANN
from keras.models import Sequential
from keras.layers import Dense

```

```python
classifier=Sequential()
```

```python
classifier.add(Dense(units=6,activation='relu',input_dim=11))
```

```python
classifier.add(Dense(units=6,activation='relu'))
```

```python
classifier.add(Dense(units=1,activation='sigmoid'))
```

```python
classifier.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
model_history=classifier.fit(x_train,y_train,batch_size=11,epochs=25,validation_split=0.33)
```
