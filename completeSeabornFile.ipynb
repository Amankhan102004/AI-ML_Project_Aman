```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df=sns.load_dataset("tips")
print(df)
```

```python
#violin plot
sns.violinplot(x="day",y="total_bill",data=df)
plt.show()
```

```python
#hue()
sns.violinplot(x="day",y="total_bill",data=df,hue="time")
plt.show()
```

```python
#linewidth()    to increase the line width
sns.violinplot(x="day",y="total_bill",data=df,hue="time",linewidth=4)
plt.show()
```

```python
# palette()   to set the color
sns.violinplot(x="day",y="total_bill",data=df,hue="time",palette="Dark2")
plt.show()
```

```python
#order    to set order wise lunch and dinner
sns.violinplot(x="time",y="total_bill",data=df,order=["Dinner","Lunch"])
plt.show()
```

```python
# saturation
sns.violinplot(x="time",y="total_bill",data=df,order=["Dinner","Lunch"],saturation=0.1)
plt.show()
```

```python
#color   To set the color of graph
sns.violinplot(x="time",y="total_bill",data=df,order=["Dinner","Lunch"],color='g')
plt.show()
```

```python
#split  and density_norm
sns.violinplot(x="day",y="total_bill",data=df,hue='sex',split=True,density_norm='width')
plt.show()
```

```python
#for vertically
sns.violinplot(x="total_bill",y="day",data=df,hue='sex',split=True,density_norm='width')
plt.show()
```

```python
#inner   this pqrameters is used to set the internal shape like point quart,stick etc
sns.violinplot(x="time",y="total_bill",data=df,order=["Dinner","Lunch"],inner="stick",color='r')
plt.show()
```

```python
#single violin plot based on single numerical column
sns.violinplot(x=df["total_bill"])
plt.show()
```

```python
sns.violinplot(x=df["tip"])
plt.show()
```

```python
sns.violinplot(y=df["size"])
plt.show()
```

Pair Plot

Pair plot   shows the relationship between multiple numecak columns

```python
sns.pairplot(df)
plt.show()
```

```python
#hue
sns.pairplot(df,hue="sex")
plt.show()
```

```python
#  between two columms relation
sns.pairplot(df,hue="sex",vars=["total_bill","tip"])
plt.show()
```

```python
#hue_order    to set the order of male and female
sns.pairplot(df,hue="sex",vars=["total_bill","tip"],palette="Accent")
plt.show()
```

```python
#hue_order()
sns.pairplot(df,hue="sex",vars=["total_bill","tip"],hue_order=["Female","Male"],palette="BuGn")
plt.show()
```

```python
#kind    change the type of graph
sns.pairplot(df,hue="sex",vars=["total_bill","tip"],kind='kde')
plt.show()
```

```python
#kind    change the type of graph
sns.pairplot(df,hue="sex",vars=["total_bill","tip"],kind='hist')
plt.show()
```

```python
# diag_kind     for set the diagonal
sns.pairplot(df,hue="sex",kind='kde',diag_kind="hist")     # kind=kde,hist,reg,scatter
plt.show()
```

```python
#marker    to set the symbol of legend

sns.pairplot(df,hue="sex",markers=["D","^"])
plt.show()
```

Strip plot
It is bsically scatter plot that differentiate different categories

```python
df=sns.load_dataset("tips")
print(df)
```

```python
sns.stripplot(x="day",y="total_bill",data=df)
plt.show()
```

```python
#hue
sns.stripplot(x="day",y="total_bill",data=df,hue="sex")
plt.show()
```

```python
#palette
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="Dark2")
plt.show()
```

```python
#linewidth
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="Dark2",linewidth=1)
plt.show()
```

```python
#edgecolor
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="Dark2",linewidth=1,edgecolor='r')
plt.show()
```

```python
#jitter    spread the dootted with spaces for better understanding
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="Dark2",linewidth=1,edgecolor='b',jitter=1)
plt.show()
```

```python
#size    this paramaeter is used to increase the size of dotted
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="Dark2",edgecolor='b',jitter=1,size=20)
plt.show()
```

```python
#marker
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="Dark2",linewidth=1,edgecolor='b',jitter=1,marker="d",size=10)
plt.show()
```

```python
#alpha
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="Dark2",linewidth=1,edgecolor='b',jitter=1,alpha=0.6,size=10)
plt.show()
```

Box plot in seaborn  --- categories the data for better understanding in the quartile and used to detect the outlier


```python
sns.set(style="darkgrid")
sns.boxplot(x="day",y="total_bill",data=df)
plt.show()
```

```python
#hue
sns.boxplot(x="day",y="total_bill",data=df,hue="sex")
plt.show()
```

```python
sns.boxplot(x="day",y="total_bill",data=df,hue="time")
plt.show()
```

```python
#color and order
sns.boxplot(x="day",y="total_bill",data=df,hue="sex",color='g',order=["Fri","sun","thu","sat"])
plt.show()
```

```python
#showmeans   this parameter is used to show the mean symbol in boxplot
sns.boxplot(x="day",y="total_bill",data=df,hue="time",showmeans=True,palette="Dark2")
plt.show()
```

```python
#meanprops
#this parameter is used to show the properties of mean like symbol
sns.boxplot(x="day",y="total_bill",data=df,hue="time",showmeans=True,palette="Dark2",meanprops={"marker":"+","markeredgecolor":"black"})
plt.show()
```

```python
#orient   for change the orientation
sns.boxplot(data=df,showmeans=True,palette="Dark2",meanprops={"marker":"+","markeredgecolor":"black"},orient="h")
plt.show()
```

```python
#only single boxplot
sns.boxplot(x=df["total_bill"])
plt.show()
```

```python
#only single boxplot
sns.boxplot(y=df["total_bill"])
plt.show()
```
