```python
import matplotlib.pyplot as plt
product=['A','B','C','D']
sales=[1000,1500,800,1200]
plt.bar(product,sales,color="orange",label='sales 2025')
plt.xlabel("product")
plt.ylabel("Sales")
plt.title("Product Sale Comparison")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
regions=['North','West','South','East']
revenue=[3420,2000,1500,1000]
plt.pie(revenue,labels=regions,autopct='%1.1f%%',colors=['Gold','purple','green','red'])
plt.title("Revenue contribution by Regions")
plt.show()
```

```python
import matplotlib.pyplot as plt
scores=[45,56,87,69,61,67,89,98,56,76,94,84,60]
plt.hist(scores,bins=5,color='purple', edgecolor='black')
plt.xlabel("score range")
plt.ylabel("number of student")
plt.show()
```

```python
import matplotlib.pyplot as plt
hour_studies=[1,2,3,4,5,6,7,8]
exam_scores=[50,55,60,65,70,75,80,85]

plt.scatter(hour_studies,exam_scores,color='red', edgecolor='black' ,marker='o',label='class A')
plt.scatter([1,2,3,4],[30,45,65,78],color='green', edgecolor='black' ,marker='o',label='class B')
plt.xlabel("Hours studies")
plt.ylabel("exam score")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[10,20,15,25]
plt.subplot(2,3,1)
plt.plot(x,y)
plt.title("Line chart")

plt.subplot(2,3,2)
plt.bar(x,y)
plt.title("Bar chart")

plt.subplot(2,3,3)
plt.pie([37,56,72,18],labels=["East","West","North","South"],autopct="%1.1f%%",colors=["green","yellow","blue","red"])
plt.title("Bar  Chart")

plt.tight_layout()
plt.savefig("subplot.png",dpi=300,bbox_inches='tight')
plt.show()

```
