```python
#image dimension
if img is not None:
  h,w,c=img.shape
  print("height:",h)
  print("width:",w)
  print("channel:",c)
else:
  print("could not load image")
```

```python
#GrayScale conversion of the image
if img is not None:
  gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
  cv2_imshow(gray)
  # cv2.waitKey(0) # Not needed with cv2_imshow
  # cv2.destroyAllWindows() # Not needed with cv2_imshow
else:
  print("could not load image")
```

```python
#Image Transformation and manipulation technique
```

```python
#Resizing and scalling the image
if img is not None:
  resized=cv2.resize(img,(200,200))
  cv2_imshow(resized)
else:
  print("No image found")
```

```python
#cropping image using slicing
cropped_image=img[100:300,200:400]
cv2_imshow(img)
cv2_imshow(cropped_image)
```

```python
#image Rotation and flipping
if img is not None:
  m=cv2.getRotationMatrix2D((w/2,h/2),45,1)
  rotated=cv2.warpAffine(img,m,(w,h))
  cv2_imshow(rotated)
else:
  print("No image found")
```

```python
if img is None:
  print("No image found")
else:
  (h,w)=img.shape[:2]
  center=(w/2,h/2)
  M=cv2.getRotationMatrix2D(center,45,2.0)
  rotated=cv2.warpAffine(img,M,(w,h))
  cv2_imshow(rotated)
```

```python
#flipped the image
if img is not None:
  flipped=cv2.flip(img,1)   #flipped horizontally
  cv2_imshow(flipped)
else:
  print("No image found")
```

```python
#flipped the image
if img is not None:
  flipped=cv2.flip(img,0)   #flipped vertically
  cv2_imshow(flipped)
else:
  print("No image found")
```

```python
#flipped the image
if img is not None:
  flipped=cv2.flip(img,-1)   #reverse flipped
  cv2_imshow(flipped)
else:
  print("No image found")
```

```python
#dfrawing the line
if img is not None:
  pt1=(50,100)
  pt2=(200,100)
  color=(255,0,0)
  thickness=4
  cv2.line(img,pt1,pt2,color,thickness)
  cv2_imshow(img)
else:
  print("No image found")
```

```python
#drawing a rectangle
if img is not None:
  pt1=(50,50)
  pt2=(450,200)
  color=(0,255,0)
  thickness=3
  cv2.rectangle(img,pt1,pt2,color,thickness)
  cv2_imshow(img)
```

```python
#drawing a circle
if img is not None:
  cv2.circle(img,(150,150),30,(0,0,255),-1)
  cv2_imshow(img)
#
```

```python
#Adding the text
cv2.putText(img,"Sun Rises in the East",(50,300),cv2.FONT_HERSHEY_SIMPLEX,1.5,(255,255,255),2)
cv2_imshow(img)
```
