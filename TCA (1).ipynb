#replacing blanks with 0 as tenure is 0 and no total charges are recorded

```python
df["TotalCharges"]=df["TotalCharges"].replace(" ","0")
df["TotalCharges"]=df["TotalCharges"].astype("float")
df.info()
```

# here we converted 0 and 1 values of senior citizen to yes or no to make it easier to understand

```python
def conv(value):
    if value==1:
        return "yes"
    else:
        return "no"
df["SeniorCitizen"]=df["SeniorCitizen"].apply(conv)
```

```python
ax=sns.countplot(x=df['Churn'])
ax.bar_label(ax.containers[0])
plt.title("count of customer by churns")
plt.show()
```

```python
gb=df.groupby("Churn").agg({"Churn":"count"})
print(gb)
```

```python
plt.pie(gb['Churn'],autopct="%1.1f%%",labels=gb.index)
plt.title("Percentage of Churned Customers",fontsize=30)
plt.show()
```

#from the given pie chart we can conclude that 26.54% of our custoners have churned out

```python
plt.figure(figsize=(4,4))
ax=sns.countplot(x="gender",data=df,hue="Churn")
for container in ax.containers:
    ax.bar_label(container)
plt.title("Churn by gender")
plt.show()
```

```python
plt.figure(figsize=(5,4))
# Create the standard side-by-side bar plot
ax = sns.countplot(x='SeniorCitizen', data=df, hue="Churn")

# Loop through ALL containers to add labels to every bar
for container in ax.containers:
    ax.bar_label(container)

plt.title("Count of customers by Senior Citizen")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create the cross-tabulation of SeniorCitizen and Churn
churn_senior_counts = pd.crosstab(df['SeniorCitizen'], df['Churn'])

# Plotting the stacked bar chart
churn_senior_counts.plot(kind='bar', stacked=True, figsize=(5, 5))

plt.title("Churn by Senior Citizen")
plt.xlabel("Senior Citizen (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.legend(title='Churn')
plt.xticks(rotation=0)
plt.show()
```

```python

# List of your columns
cols = ['PhoneService', 'MultipleLines', 'InternetService',
        'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
        'TechSupport', 'StreamingTV', 'StreamingMovies']

# Set up the figure and axes for a 3x3 grid
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(15, 12))

# Flatten the axes array for easy iteration
axes = axes.flatten()

for i, col in enumerate(cols):
    sns.countplot(data=df, x=col, ax=axes[i],hue=df["Churn"])
    axes[i].set_title(f'Count plot of {col}')
    axes[i].set_xlabel(f'{col}')
    axes[i].set_ylabel('Count')

# Adjust layout to prevent overlapping
plt.tight_layout()
plt.show()
```
