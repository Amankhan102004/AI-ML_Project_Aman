```python
import pandas as pd
import numpy as np
import cv2
```

```python
from google.colab.patches import cv2_imshow

img=cv2.imread("Natural.jpg")
cv2_imshow(img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#For to blurred the image
blurred_image=cv2.GaussianBlur(img,(9,9),0)
cv2_imshow(blurred_image)
```

```python
#median blurred
blurred=cv2.medianBlur(img,5)
cv2_imshow(blurred)
```

```python
sharpen_kernel=np.array([[-1,-1,-1],[-1,9,-1],[-1,-1,-1]])
```

```python
#Sharping the filter
sharpened_image = cv2.filter2D(img,-1,sharpen_kernel)
cv2_imshow(sharpened_image)
```

```python
#canny edge detection
edges=cv2.Canny(img,100,200)
cv2_imshow(edges)
```

```python
img=cv2.imread("Natural.jpg",cv2.IMREAD_GRAYSCALE)
edges=cv2.Canny(img,100,200)
cv2_imshow(edges)
```

```python
thresholded=cv2.threshold(img,127,255,cv2.THRESH_BINARY)[1]
cv2_imshow(thresholded)
```
