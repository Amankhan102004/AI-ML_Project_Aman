```python
import warnings
import os

# Python की सभी चेतावनियों को बंद करने के लिए
warnings.filterwarnings("ignore")

# HuggingFace / Transformers के एक्स्ट्रा लॉग्स को बंद करने के लिए
os.environ["HF_HUB_DISABLE_SYMLINKS_WARNING"] = "1"
```

```python
from langchain_huggingface import ChatHuggingFace,HuggingFacePipeline
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
llm=HuggingFacePipeline.from_model_id(
    model_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    task="text-generation",
    pipeline_kwargs=dict(
     temperature=0.5, max_new_tokens=64
     )
)
```

```python
model=ChatHuggingFace(llm=llm)
```

```python
chat_history=[]
```

```python
from langchain_core.messages import HumanMessage, AIMessage

# चैट हिस्ट्री को खाली लिस्ट से शुरू करें
chat_history = []

while True:
    user_input = input("You: ")
    if user_input.lower() == 'exit':
        break

    # 1. यूजर के इनपुट को HumanMessage ऑब्जेक्ट के रूप में हिस्ट्री में जोड़ें
    chat_history.append(HumanMessage(content=user_input))

    # 2. मॉडल को पूरी हिस्ट्री भेजें
    result = model.invoke(chat_history)

    # 3. मॉडल के जवाब को AIMessage ऑब्जेक्ट के रूप में हिस्ट्री में जोड़ें
    chat_history.append(AIMessage(content=result.content))

    print(f"AI: {result.content}")
```
